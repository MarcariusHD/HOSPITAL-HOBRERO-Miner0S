# 📊 Proyecto Hospital Obrero - Cuaderno 1: Análisis Descriptivo del Triaje
**Objetivo de Negocio:** Evaluar la distribución de niveles de urgencia en los ingresos al hospital y medir la métrica clave de negocio (Porcentaje de Triaje 4 y 5).


### 1. Carga de Librerías y Dataset
Leemos el conjunto de datos de triaje hospitalario.


```python
import pandas as pd
import numpy as np

# Cargar el dataset del hospital
df = pd.read_csv('dataset_hospital_obredo_triaje.csv')
print(f'Total de registros cargados: {len(df):,}')
df.head()
```

### 2. Filtrado y Calidad de Datos
Filtramos únicamente los registros que cuentan con un nivel de triaje válido entre 1 y 5.


```python
# Filtrar valores válidos de triaje (1, 2, 3, 4, 5)
df_valid = df[df['Nivel de triaje'].isin([1, 2, 3, 4, 5])].copy()
total_validos = len(df_valid)
print(f'Total de registros válidos procesados: {total_validos:,}')
```

### 3. Distribución de Frecuencia por Nivel de Triaje
Calculamos la cantidad absoluta de pacientes y el porcentaje relativo de atención.


```python
conteos = df_valid['Nivel de triaje'].value_counts().sort_index()
tabla_resumen = pd.DataFrame({
    'Nivel de Triaje': [f'Nivel {i}' for i in range(1, 6)],
    'Cantidad Pacientes': [conteos.get(i, 0) for i in range(1, 6)],
    'Porcentaje (%)': [(conteos.get(i, 0) / total_validos * 100) for i in range(1, 6)]
})
tabla_resumen['Porcentaje (%)'] = tabla_resumen['Porcentaje (%)'].round(2)
tabla_resumen
```

### 4. Métrica Principal de Negocio: % Triaje 4 + 5 (No Urgentes)
Calculamos la métrica principal de saturación hospitalaria.


```python
pct_no_urgente = tabla_resumen.loc[tabla_resumen['Nivel de Triaje'].isin(['Nivel 4', 'Nivel 5']), 'Porcentaje (%)'].sum()
print('='*50)
print(f'Métrica Clave de Negocio (Triaje 4 + 5): {pct_no_urgente:.2f}%')
print('='*50)
print('Conclusión: Más del 50% de las atenciones corresponden a bajas urgencias, saturando las salas de guardia.')
```
